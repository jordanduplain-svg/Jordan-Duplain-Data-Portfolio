```python
import pandas as pd

df=pd.read_csv('MCLFINI.csv',sep=';')
```

```python
df['Symptomes_psychiatriques']
```

```python
df2=df[['Sexe','Mode_entree_note','Mode_entree','Age_debut_symptomes','Symptomes_prodromaux','TCSP_prodr','Syndrome_extrapyramidal','Delai_syndrome_extrapyramidal','Symptomes_cognitifs','Date_symptomes_cognitifs','Symptomes_psychiatriques','Delai_symptomes_psychiatriques','Trouble_oculomoteur','Symptomes_moteurs','Idees_delirantes','Delai_fluctuations']]
```

```python
df2=df2.drop(df2[df2['Sexe']== '1 = féminin\n2 = masculin'].index)
```

```python
df2['Sexe']=df2['Sexe'].replace({"1" :'Feminin',"2":'Masculin'})
```

```python
df2['Symptomes_prodromaux']=df2['Symptomes_prodromaux'].replace({'1':'Oui','0':'Non'})
```

```python
df2['TCSP_prodr']=df2['TCSP_prodr'].replace({'1':'Oui','0':'Non'})
```

```python
df2['Syndrome_extrapyramidal']=df2['Syndrome_extrapyramidal'].replace({'1':'Oui','0':'Non'})
```

```python
df2['Symptomes_moteurs']=df2['Symptomes_moteurs'].replace({'1':'Oui','0':'Non'})
```

```python
df2['Idees_delirantes']=df2['Idees_delirantes'].replace({'1':'Oui','0':'Non'})
```

```python
df2['Trouble_oculomoteur']=df2['Trouble_oculomoteur'].replace({'1':'Oui','0':'Non','2':'NSP'})
```

```python
df2['Symptomes_psychiatriques']=df2['Symptomes_psychiatriques'].replace({1:'Oui',0:'Non'})
```

```python
df2
```

```python
df2.to_csv('Thesis_Aicha_Data.csv',encoding='utf-8-sig',sep=";",index=False)
```
