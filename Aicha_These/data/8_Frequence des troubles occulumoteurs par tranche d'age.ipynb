Fréquence des troubles occulumoteurs

```python
import pandas as pd
df=pd.read_csv('C:\\Users\\jorda\\PycharmProjects\\Aicha_These\\data\\Thesis_Aicha_Data V3.csv', sep=";")
df
```

```python
df2=df[['Trouble_oculomoteur','Tranche_age']]
```

```python
occulus=df2[df2['Trouble_oculomoteur']=='Oui' ]
occulus
```

```python
import plotly.express as px

counts = occulus[['Trouble_oculomoteur','Tranche_age']].value_counts().reset_index(name='counts')
counts

```

```python
fig = px.bar(
    counts,
    x='Tranche_age',
    y='counts',
    text='counts',
    title='Fréquence des troubles oculomoteurs'
)

fig.show()
```
