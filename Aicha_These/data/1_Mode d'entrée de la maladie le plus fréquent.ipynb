Savoir quel est le mode d'entrée de la maladie le plus fréquent dans ces groupes là

```python
import pandas as pd
import plotly.express as px
```

```python
df=pd.read_csv('C:\\Users\\jorda\\PycharmProjects\\Aicha_These\\data\\Thesis_Aicha_Data.csv', sep=";")
```

```python
df
```

```python
mode_entree=df[['Mode_entree_note','Mode_entree']]
mode_entree
```

```python
df['Mode_entree'] = df['Mode_entree'].replace({
    1: "Trouble Cognitfs",
    2: "Trouble Psychatrique",
    3: "Trouble Moteur",
    4: "Trouble Psycharique et Cognitifs"})
df['Nom Mode_entree']=df['Mode_entree']
```

```python
df.to_csv('Thesis_Aicha_Data V2.csv', encoding='utf-8', sep=';',index=False)
```

```python
df['Mode_entree'].dropna()
```

```python
counts=df['Mode_entree'].value_counts().reset_index(name='counts')
counts.columns=['Mode_entree','counts']
```

```python
counts
```

```python
counts.plot(x='Mode_entree', y='counts', title="Mode d'entrée des patients", kind='pie')
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('_mpl-gallery-nogrid')

counts = df['Mode_entree'].value_counts()
labels = counts.index

fig, ax = plt.subplots()
ax.pie(counts, labels=labels, autopct='%1.1f%%', startangle=90)
plt.show()
```

```python
fig = px.histogram(df['Mode_entree'], x="Mode_entree",nbins=10)
fig.show()
```
