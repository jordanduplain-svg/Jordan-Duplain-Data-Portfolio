Classe d'age ou les symptomes Prodormaux sont les plus fréquents

```python
import pandas as pd
import numpy as np

df=pd.read_csv('C:\\Users\\jorda\\PycharmProjects\\Aicha_These\\data\\Thesis_Aicha_Data V3.csv', sep=";")

df
```

```python
df2=df[['Symptomes_prodromaux','Tranche_age']]
```

```python
df2=df2.dropna()
```

```python
df2
```

```python
prodormaux=df2[df2['Symptomes_prodromaux']=='Oui']
prodormaux
```

```python
counts=prodormaux[['Symptomes_prodromaux','Tranche_age']].value_counts().reset_index(name='counts')

counts
```

```python
import plotly.express as px
fig = px.pie(
    counts,
    values='counts',
    names='Tranche_age',
    title= "Symptome Prodromaux par tranche d'age"
)
fig.show()
```
